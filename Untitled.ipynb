```python
import json
import pathlib

import numpy as np
import pandas as pd

import matplotlib.pyplot as plt
import matplotlib.ticker as mticker
from matplotlib.lines import Line2D
%matplotlib inline
```

```python
RHO_YLIM = None
BETA_YLIM = (0, 10)
TAU_YLIM = (-30, 30)
MU_YLIM = None


def panel(ax, df, key, title, ylim=None, log=False):
    for l in range(6):
        for h in range(6):
            (ax.semilogy if log else ax.plot)(df["train.iter"], df[f"diagnose.head({l}, {h}).{key}"], "-", color=colors[l])
    ax.set_xlabel("iter")
    ax.set_title(title)
    if ylim:
        ax.set_ylim(*ylim)


def plots(filename: str, subtitle: str, n_layers: int = 6) -> None:

    with open(filename, 'r') as f:
        data = [json.loads(line) for line in f]
    
    df = pd.json_normalize(data)
    
    colors = plt.cm.viridis(np.linspace(0, 1, n_layers))
    fig, axs = plt.subplots(1, 5, figsize=(13, 3.4), constrained_layout=True)
    
    axs[0].semilogy(df["train.iter"], df["train.loss"], '-k')
    axs[0].set_xlabel("iter")
    axs[0].set_ylim(0.5, 5) 
    axs[0].set_title(f"loss ({subtitle})")

    panel(axs[1], df, "rho", f"rho ({subtitle})", ylim=RHO_YLIM, log=True)
    panel(axs[2], df, "con_drift", f"beta ({subtitle})", ylim=BETA_YLIM)
    panel(axs[3], df, "tau", f"tau ({subtitle})", ylim=TAU_YLIM)
    panel(axs[4], df, "mu", f"mu ({subtitle})", ylim=MU_YLIM)
    
    handles = [Line2D([0], [0], color=colors[l], lw=1.5, label=f"layer {l}") for l in range(n_layers)]
    fig.legend(handles=handles, loc="outside right center", frameon=False)

    fig.savefig(pathlib.Path(filename).with_suffix(".png"), dpi=300, bbox_inches="tight")


def allplots(data: dict[str, str]) -> None:
    
    colors = plt.cm.viridis(np.linspace(0, 1, 6))
    fig, axs = plt.subplots(len(data), 5, figsize=(13, 3.4 * len(data)), constrained_layout=True)

    for row, (filename, subtitle) in enumerate([(k, v) for k, v in data.items()]):
        
        with open(filename, 'r') as f:
            data = [json.loads(line) for line in f]
        
        df = pd.json_normalize(data)
        
        axs[row, 0].semilogy(df["train.iter"], df["train.loss"], '-k')
        axs[row, 0].set_xlabel("iter")
        axs[row, 0].set_ylim(0.5, 5) 
        axs[row, 0].set_title(f"loss ({subtitle})")
    
        panel(axs[row, 1], df, "rho", f"rho ({subtitle})", ylim=RHO_YLIM, log=True)
        panel(axs[row, 2], df, "con_drift", f"beta ({subtitle})", ylim=BETA_YLIM)
        panel(axs[row, 3], df, "tau", f"tau ({subtitle})", ylim=TAU_YLIM)
        panel(axs[row, 4], df, "mu", f"mu ({subtitle})", ylim=MU_YLIM)
    
    handles = [Line2D([0], [0], color=colors[l], lw=1.5, label=f"layer {l}") for l in range(6)]
    fig.legend(handles=handles, loc="outside right center", frameon=False)
    fig.savefig("plots.png", dpi=300, bbox_inches="tight")
```

```python
plots("stats.jsonl", "sgd, wd=0.4,lr=4e-4")
```

```python
plots("results/sgd-wo-weight-decay.jsonl", "sgd, wd=0.0")
plots("results/sgd-w-weight-decay.jsonl", "sgd, wd=0.1")
plots("results/adamw.jsonl", "adamw, wd=0.1")
```

```python
allplots({
    "results/sgd-wo-weight-decay.jsonl": "sgd, wd=0.0",
    "results/sgd-w-weight-decay.jsonl": "sgd, wd=0.1",
    "results/adamw.jsonl": "adamw, wd=0.1"
})
```
